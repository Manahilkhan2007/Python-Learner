Question 1-Even or Odd

```
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

Question 2-Simple calculater

```
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

if operator == "+":
    print("Result:", num1 + num2)
elif operator == "-":
    print("Result:", num1 - num2)
elif operator == "*":
    print("Result:", num1 * num2)
elif operator == "/":
    print("Result:", num1 / num2)
else:
    print("Invalid operator")
```

Question 3-Student marks

```
marks1 = float(input("Enter marks of subject 1: "))
total1 = float(input("Enter total marks of subject 1: "))

marks2 = float(input("Enter marks of subject 2: "))
total2 = float(input("Enter total marks of subject 2: "))

marks3 = float(input("Enter marks of subject 3: "))
total3 = float(input("Enter total marks of subject 3: "))

total_marks = marks1 + marks2 + marks3
total_possible = total1 + total2 + total3

percentage = (total_marks / total_possible) * 100

print("Total:", total_marks)
print("Percentage:", percentage, "%")

if percentage >= 50:
    print("Pass")
else:
    print("Fail")
```

Question 4-Favorite fruit

```
fruits = ["Apple", "Mango", "Banana", "Orange", "Grapes"]

fruit = input("Enter a fruit: ")

if fruit in fruits:
    print("Fruit exists in the list.")
else:
    print("Fruit does not exist in the list.")
```
